```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('netflix_cleaned.csv', encoding='latin-1')

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
content_counts = df['type'].value_counts()

print("Content Distribution:")
print(content_counts)

total_movies = content_counts.get('Movie', 0)
total_tv_shows = content_counts.get('TV Show', 0)

print("\nTotal Movies:", total_movies)
print("Total TV Shows:", total_tv_shows)

print("Total Content:", total_movies + total_tv_shows)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='type')

plt.title('Netflix Movies vs TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Total Count')

plt.savefig('content_distribution.png',
            dpi=300, bbox_inches='tight')

plt.show()
```

```python
labels = content_counts.index
values = content_counts.values

plt.figure(figsize=(7, 7))

plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Netflix Content Percentage Distribution')

plt.savefig('content_percentage.png',
            dpi=300, bbox_inches='tight')

plt.show()
```

```python
total = total_movies + total_tv_shows

movie_percentage = (total_movies / total) * 100
tv_percentage = (total_tv_shows / total) * 100

print("KEY FINDINGS")
print("----------------------")

print("Total Movies:", total_movies)
print("Total TV Shows:", total_tv_shows)

print(f"Movies Percentage: {movie_percentage:.2f}%")
print(f"TV Shows Percentage: {tv_percentage:.2f}%")

if total_movies > total_tv_shows:
    print("Movies are more available on Netflix.")
else:
    print("TV Shows are more available on Netflix.")
```
